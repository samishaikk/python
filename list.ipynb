### List comprehension

Basic syntax [expression for item in iterable]

with conditional logic [expression for item in iterable if condition]

Nested list comprehension [iterating through two different iterables]

```python
square = []

for num in range(10):
    if num%2==0:
        square.append(num**2)

print(square)
```

```python
square = [num**2 for num in range(10)]
print(square)
```

```python
# with condition 
even_square = [num**2 for num in range(32) if num%2==0]
print(even_square)
```

```python
odd_square = [num**2 for num in range(10) if num%2!=0]
print(odd_square)
```

```python
list1 = [1,2,3]
list2 = ['a','b','c']

for x in list1:
    for y in list2:
        print((x,y))    
        

pair = [(x,y) for x in list1 for y in list2]
print(pair)
```

```python
# list comprehension with function calls

words = ["hello", "world", "python"]
lengths = [len(word) for word in words]
print(lengths)
```
