### Dictionaries

- Unordered collection of items
- Store data in key-value pairs
- keys must be unique and immutable eg: string, numbers, or tuples
- Values can be of any type

```python
empty_dict = {}
type(empty_dict)
```

```python
empty_dict = dict()
type(empty_dict)
```

```python
Student = {"name":"sami","age":33,"grade":9.4}
Student
```

```python
Student = {"name":"sami","age":32,"name":32}
Student
```

```python
Student = {"name":"sami","age":33,"name":"sam"}
print(Student)
```

```python
#  accessing dictionary elements

print(Student['name'])
print(Student['age'])
print("\n")
# accessing using get() method
print(Student.get('name'))
print(Student.get('age'))
print(f"age is {Student.get('age')}")
```


- Modifying dictionary elements
- Dictionaries are mutable, so you can add, update or delete elements

```python
Student
```

```python
Student["age"] = 34
Student["address"] = "India"
Student["grade"] = "A"
print(Student)
```

```python
del Student["grade"]
print(Student)
```

```python
# Dictionary methods

keys = Student.keys()
print(keys)
values = Student.values()
print(values)

items = Student.items()
print(items)
```

```python
# Shallow copy

studentcopy = Student
print(studentcopy)
print(Student)
```

```python
Student["name"] = "sam4"
print(Student)
print(studentcopy)
```

```python
studentcopy1 = Student.copy() # shallow copy

print(studentcopy1)
print(Student)
```

```python
Student["name"] = "sam2"

print(studentcopy)
print(studentcopy1)
```

```python
studencopy = Student
print(studencopy)

shallowcopy = Student.copy()
print(shallowcopy)
```

```python
Student["name"] = "sam3"

print(studencopy)
print(shallowcopy)
```

```python
# iterating over dictionaries
# you can use loops to iterate over dictionaries, keys, values or items

# iterating overy keys
for keys in Student.keys():
    print(keys)
```

```python
for values in Student.values():
    print(values)
```

```python
# iterate over key value pairs
for key,value in Student.items():
    print(f"{key }: {value}")
```

```python
Students = {
    "studen1":{"name":"sami","age":22},
    "studen2":{"name":"peter","age":33}
}
print(Students)
```

```python
print(Students["studen2"]['age'])
print(Students['studen1']['age'])
```

```python
# Students.items()
for student_info,student_id in Students.items():
    print(student_info,student_id)
    for key,value in student_id.items():
        print(key,value)
    
```

```python
# dictionary comprehension

squares = {x: x**2 for x in range(3)}
squares
```

```python
# conditional dictionary comprehension

squares_withcondtion = {x: x**2 for x in range(10) if x%2==0}
squares_withcondtion
```

```python
# Examples

# Use dictionary to count the frequency of elements in the list 

numlist = [1,2,2,4,5,55,23,44,44,3,2,3,4,6,777,55]
# frequency = numlist.count(2)
# print(f'no of 2s is {frequency}')
frequency = {}

for number in numlist:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
    
print(frequency)
```
