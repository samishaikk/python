```python
# break and continue
# break -> exit the loop
# continue -> skip the current iteration and move to next iteration

for i in range(10):
    if i==5:
        continue
    print(i)

    # if i==6:
    #     break
    # print(i)
```

```python
for i in range (11):
    if i%2==0:
        continue    
    print(i)
            
```

```python
# nested loops

for i in range (3):
    for j in range(2):
        print(f'i={i},j={j}')
```

```python
n = 10
sum = 0
count = 1

while count<=n:
    sum += count
    count += 1

print(f'sum of first {n} natural numbers is {sum}')
```

```python
n = 100

for num in range(n+1):
    if num>1:
        for i in range(2,num):
            if (num%i)==0:
                break
        else:
            print(num)
```
