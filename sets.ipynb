```python
# set methods

set1 = {1, 2, 3, 4, 5}
set2 = {3, 4, 5}
```

```python
set1.issubset(set2)  # check if set1 elements are in set2
```

```python
set1.issuperset(set2)  # check if set1 has all elements of set2
```

```python
set1.symmetric_difference(set2)  # return unique elements in both sets
```
