
## Variables

- Variables store memory in for the value.
- Variable names can be created with A-Z,a-z,0-9 and _.
- No special characters are allowed.
- A variable should always start with a character or _.
- It cannot start with a number.
- Spaces are not allowed in variable naming.

### Basic syntax rules in python

```python
# Case sensitive

name = "sami"
Name = "samii"

print(name)
print(Name)
```

```python
# Indentation is important
# Python uses identation to define blocks of code. Consistent indentation is crucial for the code to run correctly.
age = int(input("Enter your age: "))
if age > 18:
    print("You are an adult.")
    print("You can vote.")
```

```python
# Line continuation
total = 1 + 2 + 3 + \
        4 + 5 + 6
    
print("Total:", total)
```

```python
# Multiple statements on a single line
x = 5; y = 10; z = x + y
print("Sum:", z)
```

```python
# Understanding semantics
# variable assignment
a = 10 
name = "sami"
type(a), type(name)
```

```python
variable1 = 10;variable = 'sam'
print(type(variable1), type(variable))
```
