#### Ex1 -> Temp conversion

```python
def convert_temperature(temp,unit):
    if unit=='C':
        return temp* 9/5 + 32
    elif unit=='F':
        return (temp-32)* 5/9
    else:
        None

print(convert_temperature(25,'C'))
print(convert_temperature(98.6, 'F'))
```

#### Ex2 -> Password strength checker

```python
def is_strong_password(password):
    if len(password) < 8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in '!@#$%^&*()_+' for char in password):
        return False
    return True
print(is_strong_password('weakpassword'))
print(is_strong_password('Sawei34#'))
```

#### Ex3 -> Calculate total cost of items in shopping cart

```python
def calculate_total_cost(cart):
    total_cost = 0
    
    for item in cart:
        total_cost+=item['price']*item['quantity']
        
    return total_cost

cart = [{'name':'apple','price':0.4,'quantity':5},
        {'name':'banana','price':0.3,'quantity':6},
        {'name':'orange','price':0.7,'quantity':3}]

# calling function

total_cost = calculate_total_cost(cart)
print(total_cost)
```

#### Ex4 -> check if string is palindrome

```python
def is_palidrome(givenstring):
    
    givenstring= givenstring.lower().replace(" ","")
    
    if givenstring==givenstring[::-1]:
        return 'palindrome'
    else:
        return 'not palindrome'
    
print(is_palidrome('sami'))
print(is_palidrome('a man a plan acanal panama'))
```

#### Ex5 -> Calculate factorial of number using recursion

```python
def factorial(n):
    if n==0:
        return 1
    else:
        return n * factorial(n-1)
    
print(factorial(5))
```

#### Ex6 -> write a Function to read a file and count frequency of each word

```python
def count_word_frequency(file_path):
    word_count={}
    
    try:
        with open(file_path,'r') as file:
            for line in file:
                words = line.split()
                for word in words:
                    word = word.lower().strip('.,!?;:"\'')
                    word_count[word] = word_count.get(word,0)+1
        return word_count
    
    except:            
        return 'error reading'

word_frequency = count_word_frequency(file_path='sample.txt')
print(word_frequency)
        
```

#### Ex7 -> validate email address

```python
import re

def is_valid_email(email):
    pattern = r'^[a-zA-Z0-9_.+-]+@[a-zA-Z0-9-]+\.[a-zA-z0-9-.]+$'
    return re.match(pattern,email) is not None

print(is_valid_email('abd@gmail.com'))
print(is_valid_email('invalid/email@gmail.com'))
```
