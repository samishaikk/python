```python
#  why functions?

def even_or_odd(num):
    if num%2==0:
        print('even')
    else:
        print('odd')

# num = int(input('enter input: '))
# even_or_odd(num)

even_or_odd(num=33)
```

```python
# function with multiple parameters

def add(a,b):
    return a+b

# add(a=3,b=5)

result = add(3,9)
print(result)
    
```

```python
# default parameters

def greet(name): # by default it'll take guest if not given
    print(f'Hello {name} welcome to mehfil')

greet('sami')
```

```python
# variable length arguments
# positional and keywords arguments

def print_numbers(*sami):
    for number in sami:
        print(number)
```

```python
print_numbers(1,2,3,4,5,"sami")
```

```python
def print_numbers(*args):
    for number in args:
        print(number)

print_numbers(1,2,3,4,5,'sami')
```

```python
# keyword arguments

# All the parameters will be in the form of key-value pairs

def print_details(*args,**kwargs):
    for val in args:
        print(f'position argument:{val}')
    for key,value in kwargs.items():
        print(f'{key}:{value}')
```

```python
print_details(1,2,3,name='sami',age=22,country='India')
```

```python
# return multiple parameters

def multiply(a,b):
    return a*b,a

multiply(2,3)
```
