```python
def gener_squaree(n):
    result = []
    for _ in range(n):
        result.append('*'*n)
    return result

gener_squaree(3)
```

```python
def square_of_side(n):
    # Return a list of strings, each containing n asterisks
    return ['*' * n for _ in range(n)]


# Example test cases
print(square_of_side(3))  
```

```python
# celsius to farenheit

# celsiusinput = float(input())

def celsius_to_fahrenheit(C):
    
    Fahrenheitt = (9/5 * C) + 32
    
    return Fahrenheitt
    
    
celsius_to_fahrenheit(C=float(input()))
```

```python
# adding elements in the list 

numbers = list(map(int,input('enter your list:').split()))

def sum_list(numbers):
    
    result = 0
    
    for i in numbers:
        result+=i
        
    return result

sum_list(numbers)
```
