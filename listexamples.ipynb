### Example 1 -> manage a to do list

- create a To Do list to keep track of tasks

```python
to_do_list = ["wake up early","exercise","work","communicate"]

# adding to task
to_do_list.append("spend time with your family")
to_do_list.append("Take time for yourself")

# checking if task is in list

if "exercise" in to_do_list:
    print("don't forget to exercise")
    
# print(to_do_list)
for task in to_do_list:
    print(f'-{task}')
    
```

### Exmple 2 -> Organizing student grades

- create a list to store and calculate average grades for studnts

```python
grdes = [84,88,99,86]
grdes.append(90)

# calculating average grde
average_grade = sum(grdes)/len(grdes)
print(f'Average grades is: {average_grade:.2f}')

# finding highest and lowest
highestgrade = max(grdes)
lowestgrade = min(grdes)
print(highestgrade,"\n",lowestgrade)
```

### Example 4 -> collecting user feedback

- Use a list to collect and analyze user feedback.

```python
# Collecting user feedback
feedback = ['great','could be better','excellent','not that great']

# appending
feedback.append('not happy with service')

# counting specific feedback
positivefeedback = sum(1 for comment in feedback if "great" in comment.lower() or "excellent"  in comment.lower())

print(f"postive feedback count:{positivefeedback}")

# printing all of the feedback

print("all user feedbacks")

for comment in feedback:
    print(f'-{comment}')
```
